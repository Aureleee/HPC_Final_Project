```python
def is_cyclic(graph):
    """
    Returns True if the directed graph contains a cycle, otherwise False.
    graph: dictionary where
           key = vertex
           value = list of neighboring vertices
    """
    state = {vertex: 0 for vertex in graph}

    # Add vertices that appear only in adjacency lists
    for neighbors in graph.values():
        for vertex in neighbors:
            if vertex not in state:
                state[vertex] = 0

    def dfs(vertex):
        state[vertex] = 1  # visiting

        for neighbor in graph.get(vertex, []):
            if state[neighbor] == 1:
                return True
            if state[neighbor] == 0:
                if dfs(neighbor):
                    return True

        state[vertex] = 2  # visited
        return False

    for vertex in state:
        if state[vertex] == 0:
            if dfs(vertex):
                return True

    return False


# Test cases
test_graphs = {
    "Acyclic graph": {
        'A': ['B'],
        'B': ['C'],
        'C': []
    },
    "Simple cycle": {
        'A': ['B'],
        'B': ['C'],
        'C': ['A']
    },
    "Disconnected graph with cycle": {
        'A': ['B'],
        'B': [],
        'C': ['D'],
        'D': ['C']
    },
    "Self-loop": {
        'A': ['A'],
        'B': ['C'],
        'C': []
    },
    "Empty graph": {}
}

for name, graph in test_graphs.items():
    print(f"{name}: {is_cyclic(graph)}")
```
