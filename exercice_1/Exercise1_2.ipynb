```python
from collections import deque

def is_bipartite(graph):
    color = {}

    for start in graph:
        if start not in color:
            color[start] = 0
            queue = deque([start])

            while queue:
                u = queue.popleft()
                for v in graph[u]:
                    if v not in color:
                        color[v] = 1 - color[u]
                        queue.append(v)
                    elif color[v] == color[u]:
                        return False
    return True


def is_empty(graph):
    for vertex in graph:
        if len(graph[vertex]) > 0:
            return False
    return True


def is_complete(graph):
    n = len(graph)
    for vertex in graph:
        if len(graph[vertex]) != n - 1:
            return False
    return True


test_graphs = {
    "Graph 1": {
        'A': ['B'],
        'B': ['A', 'C'],
        'C': ['B']
    },
    "Graph 2": {
        'A': [],
        'B': [],
        'C': []
    },
    "Graph 3": {
        'A': ['B', 'C'],
        'B': ['A', 'C'],
        'C': ['A', 'B']
    },
    "Graph 4": {
        'A': ['B', 'D'],
        'B': ['A', 'C'],
        'C': ['B', 'D'],
        'D': ['A', 'C']
    }
}

for name, graph in test_graphs.items():
    print(name)
    print("Bipartite:", is_bipartite(graph))
    print("Empty:", is_empty(graph))
    print("Complete:", is_complete(graph))
    print()
```
